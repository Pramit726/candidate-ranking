```python
import json
from collections import Counter

jsonl_path = '../data/candidates_excluding_sample.jsonl'

WRONG_TITLE_KEYWORDS = [
    "marketing", "sales", "hr ", "human resource", "recruiter", "talent acquisition",
    "finance", "accountant", "accounting", "business development", "bd manager",
    "supply chain", "logistics", "operations manager", "project manager",
    "product manager",
    "graphic design", "ux designer", "ui designer", "visual design",
    "content writer", "copywriter", "seo", "social media",
    "civil engineer", "mechanical engineer", "electrical engineer",
    "hardware engineer", "embedded", "firmware",
    "computer vision", "cv engineer",
    "speech", "audio", "robotics", "autonomous",
    "teacher", "professor", "lecturer", "faculty",
    "doctor", "physician", "nurse", "pharmacist",
    "lawyer", "legal", "compliance",
    "customer success", "customer support", "support engineer",
]

# ── Consulting firms from the JD's explicit disqualifier list ─────────────────
PURE_CONSULTING_FIRMS = {
    "tcs", "tata consultancy", "infosys", "wipro", "accenture",
    "cognizant", "capgemini", "hcl", "tech mahindra", "mphasis",
    "hexaware", "mindtree", "ltimindtree", "l&t infotech",
    "persistent systems", "niit technologies", "mastech",
    "kforce", "syntel", "unison",
}

INDIA_VARIANTS = {"india", "in", "ind"}

def matches_any(text, keywords):
    return any(kw in text.lower() for kw in keywords)

def is_consulting_firm(company_name):
    c = company_name.lower().strip()
    return any(firm in c for firm in PURE_CONSULTING_FIRMS)

def is_pure_consulting_career(career_history):
    """
    Returns True only if EVERY role in career history is at a consulting firm.
    If even one role is at a product company → keep the candidate (JD says that's fine).
    """
    if not career_history:
        return False
    return all(is_consulting_firm(role.get("company", "")) for role in career_history)

# ── Thresholds ────────────────────────────────────────────────────────────────
MIN_YOE = 3
MAX_YOE = 15

# ── Counters & buckets ────────────────────────────────────────────────────────
country_counter     = Counter()
title_counter       = Counter()
yoe_distribution    = Counter()
consulting_firm_counter = Counter()  # frequency of consulting firms seen

non_india_ids        = []
wrong_title_ids      = []
too_junior_ids       = []
too_senior_ids       = []
honeypot_skill_ids   = []
null_skills_ids      = []
pure_consulting_ids  = []

total = 0
passed = []

with open(jsonl_path) as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        c = json.loads(line)
        total += 1
        p             = c["profile"]
        cid           = c["candidate_id"]
        career        = c.get("career_history") or []

        country = p.get("country", "").strip()
        title   = p.get("current_title", "").strip()
        yoe     = p.get("years_of_experience", 0) or 0
        skills  = c.get("skills") or []

        bucket = (int(yoe) // 2) * 2
        yoe_distribution[bucket] += 1

        exclude_reasons = []

        # ── Filter 1: country ─────────────────────────────────────────────────
        country_counter[country] += 1
        if country.lower() not in INDIA_VARIANTS:
            willing = c.get("redrob_signals", {}).get("willing_to_relocate", False)
            if not willing:
                non_india_ids.append((cid, country))
                exclude_reasons.append("non_india")

        # ── Filter 2: wrong title ─────────────────────────────────────────────
        title_counter[title] += 1
        if matches_any(title, WRONG_TITLE_KEYWORDS):
            wrong_title_ids.append((cid, title))
            exclude_reasons.append("wrong_title")

        # ── Filter 3: years of experience extremes ────────────────────────────
        if yoe < MIN_YOE:
            too_junior_ids.append((cid, yoe))
            exclude_reasons.append("too_junior")
        elif yoe > MAX_YOE:
            too_senior_ids.append((cid, yoe))
            exclude_reasons.append("too_senior")

        # ── Filter 4: null / empty skills ────────────────────────────────────
        if not skills:
            null_skills_ids.append(cid)
            exclude_reasons.append("null_skills")

        # ── Filter 5: honeypot skill signal ───────────────────────────────────
        if skills:
            expert_skills = [s for s in skills if s.get("proficiency") == "expert"]
            if expert_skills and all(
                (s.get("duration_months") or 0) == 0 for s in expert_skills
            ):
                honeypot_skill_ids.append(
                    (cid, [(s["name"], s.get("duration_months")) for s in expert_skills])
                )
                exclude_reasons.append("honeypot_skill")

        # ── Filter 6: pure consulting career ─────────────────────────────────
        # count firm appearances regardless (for the frequency report)
        for role in career:
            company = role.get("company", "")
            if is_consulting_firm(company):
                consulting_firm_counter[company] += 1

        if is_pure_consulting_career(career):
            companies = [r.get("company", "") for r in career]
            pure_consulting_ids.append((cid, companies))
            exclude_reasons.append("pure_consulting")

        if not exclude_reasons:
            passed.append(cid)

# ── Results ───────────────────────────────────────────────────────────────────

print("=" * 60)
print(f"TOTAL CANDIDATES: {total:,}")
print("=" * 60)

# YOE distribution
print("\n── YEARS OF EXPERIENCE DISTRIBUTION ──")
print(f"  {'YOE bucket':<15} {'count':>7}   bar")
for bucket in sorted(yoe_distribution):
    cnt = yoe_distribution[bucket]
    bar = "█" * (cnt // 200)
    label = f"{bucket}-{bucket+1}"
    marker = " ← MIN cutoff" if bucket == int(MIN_YOE // 2) * 2 else \
             " ← MAX cutoff" if bucket == int(MAX_YOE // 2) * 2 else ""
    print(f"  {label:<15} {cnt:>7,}   {bar}{marker}")
print(f"\n  Too junior (< {MIN_YOE} yoe):  {len(too_junior_ids):,}")
print(f"  Too senior (> {MAX_YOE} yoe): {len(too_senior_ids):,}")

# Null skills
print(f"\n── NULL / EMPTY SKILLS ──")
print(f"  Candidates with no skills listed: {len(null_skills_ids):,}")

# Honeypot skills
print(f"\n── HONEYPOT SKILL SIGNAL ──")
print(f"  Candidates with ALL expert skills at duration_months=0: {len(honeypot_skill_ids):,}")
print("  Sample (first 10):")
for cid, skills_info in honeypot_skill_ids[:10]:
    print(f"    {cid}  →  {skills_info}")

# Consulting firms
print(f"\n── CONSULTING FIRM FREQUENCY (all appearances in career history) ──")
for firm, cnt in consulting_firm_counter.most_common():
    print(f"  {firm:<40} {cnt:>6,}")
print(f"\n  Candidates with PURE consulting career: {len(pure_consulting_ids):,}")
print("  Sample (first 10):")
for cid, companies in pure_consulting_ids[:10]:
    print(f"    {cid}  →  {companies}")

# Combined
excluded_ids = (
    set(cid for cid, _ in non_india_ids)
    | set(cid for cid, _ in wrong_title_ids)
    | set(cid for cid, _ in too_junior_ids)
    | set(cid for cid, _ in too_senior_ids)
    | set(null_skills_ids)
    | set(cid for cid, _ in honeypot_skill_ids)
    | set(cid for cid, _ in pure_consulting_ids)
)

print(f"\n── COMBINED EXCLUSION SUMMARY ──")
print(f"  Non-India (no relocation):          {len(non_india_ids):>6,}")
print(f"  Wrong title:                        {len(wrong_title_ids):>6,}")
print(f"  Too junior (< {MIN_YOE} yoe):              {len(too_junior_ids):>6,}")
print(f"  Too senior (> {MAX_YOE} yoe):             {len(too_senior_ids):>6,}")
print(f"  Null/empty skills:                  {len(null_skills_ids):>6,}")
print(f"  Honeypot skill signal:              {len(honeypot_skill_ids):>6,}")
print(f"  Pure consulting career:             {len(pure_consulting_ids):>6,}")
print(f"  ─────────────────────────────────────────")
print(f"  Union (unique excluded):            {len(excluded_ids):>6,}  ({len(excluded_ids)/total*100:.1f}%)")
print(f"  Remaining after ALL filters:        {len(passed):>6,}")
```

```python
import json
from collections import Counter

jsonl_path = '../data/candidates_excluding_sample.jsonl'

WRONG_TITLE_KEYWORDS = [
    "marketing", "sales", "hr ", "human resource", "recruiter", "talent acquisition",
    "finance", "accountant", "accounting", "business development", "bd manager",
    "supply chain", "logistics", "operations manager", "project manager",
    "product manager",
    "graphic design", "ux designer", "ui designer", "visual design",
    "content writer", "copywriter", "seo", "social media",
    "civil engineer", "mechanical engineer", "electrical engineer",
    "hardware engineer", "embedded", "firmware",
    "computer vision", "cv engineer",
    "speech", "audio", "robotics", "autonomous",
    "teacher", "professor", "lecturer", "faculty",
    "doctor", "physician", "nurse", "pharmacist",
    "lawyer", "legal", "compliance",
    "customer success", "customer support", "support engineer",
]

# ── Consulting firms from the JD's explicit disqualifier list ─────────────────
PURE_CONSULTING_FIRMS = {
    "tcs", "tata consultancy", "infosys", "wipro", "accenture",
    "cognizant", "capgemini", "hcl", "tech mahindra", "mphasis",
    "hexaware", "mindtree", "ltimindtree", "l&t infotech",
    "persistent systems", "niit technologies", "mastech",
    "kforce", "syntel", "unison",
}

INDIA_VARIANTS = {"india", "in", "ind"}

def matches_any(text, keywords):
    return any(kw in text.lower() for kw in keywords)

def is_consulting_firm(company_name):
    c = company_name.lower().strip()
    return any(firm in c for firm in PURE_CONSULTING_FIRMS)

def is_pure_consulting_career(career_history):
    """
    Returns True only if EVERY role in career history is at a consulting firm.
    If even one role is at a product company → keep the candidate (JD says that's fine).
    """
    if not career_history:
        return False
    return all(is_consulting_firm(role.get("company", "")) for role in career_history)

def is_job_hopper(career_history):
    """
    Returns (bool, reason_str) 
    Flags if BOTH conditions are true:
      1. Average tenure across all roles < 18 months
      2. 3+ company switches (distinct companies)
    """
    if not career_history or len(career_history) < 2:
        return False, ""

    # ── condition 1: average tenure ──────────────────────────────────────────
    durations = [r.get("duration_months") or 0 for r in career_history]
    avg_tenure = sum(durations) / len(durations)

    # ── condition 2: distinct company switches ────────────────────────────────
    companies = [r.get("company", "").strip().lower() for r in career_history]
    unique_companies = set(companies)
    switches = len(unique_companies)

    if avg_tenure < 18 and switches >= 3:
        return True, f"avg_tenure={avg_tenure:.1f}mo, switches={switches}"

    return False, ""


def is_title_chaser(career_history):
    """
    Returns (bool, reason_str)
    Detects Senior → Staff → Principal progression across SHORT stints.
    Flags if:
      - 2+ title-level jumps detected (junior→mid, mid→senior, senior→staff, etc.)
      - AND each of those jumping roles lasted < 18 months
    """
    TITLE_LEVELS = {
        # level 1
        "junior": 1, "associate": 1, "intern": 1, "trainee": 1,
        # level 2
        "engineer": 2, "developer": 2, "analyst": 2, "scientist": 2,
        # level 3
        "senior": 3, "sr.": 3, "sr ": 3, "lead": 3,
        # level 4
        "staff": 4, "specialist": 4, "expert": 4,
        # level 5
        "principal": 5, "architect": 5,
        # level 6
        "director": 6, "vp": 6, "head": 6, "manager": 6,
    }

    def title_level(title):
        t = title.lower()
        matched = [lvl for kw, lvl in TITLE_LEVELS.items() if kw in t]
        return max(matched) if matched else 0

    # sort by start_date ascending to get chronological order
    sorted_career = sorted(
        career_history,
        key=lambda r: r.get("start_date") or "0000-00-00"
    )

    jumps = 0
    prev_level = 0
    for role in sorted_career:
        level = title_level(role.get("title", ""))
        duration = role.get("duration_months") or 0
        if level > prev_level and prev_level != 0 and duration < 18:
            jumps += 1
        prev_level = max(prev_level, level)

    if jumps >= 2:
        titles = [r.get("title", "") for r in sorted_career]
        return True, f"quick_jumps={jumps}, titles={titles}"

    return False, ""

# ── Thresholds ────────────────────────────────────────────────────────────────
MIN_YOE = 3
MAX_YOE = 15

# ── Counters & buckets ────────────────────────────────────────────────────────
country_counter     = Counter()
title_counter       = Counter()
yoe_distribution    = Counter()
consulting_firm_counter = Counter()  # frequency of consulting firms seen

non_india_ids        = []
wrong_title_ids      = []
too_junior_ids       = []
too_senior_ids       = []
honeypot_skill_ids   = []
null_skills_ids      = []
pure_consulting_ids  = []
job_hopper_ids      = []
title_chaser_ids    = []
total = 0
passed = []

with open(jsonl_path) as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        c = json.loads(line)
        total += 1
        p             = c["profile"]
        cid           = c["candidate_id"]
        career        = c.get("career_history") or []

        country = p.get("country", "").strip()
        title   = p.get("current_title", "").strip()
        yoe     = p.get("years_of_experience", 0) or 0
        skills  = c.get("skills") or []

        bucket = (int(yoe) // 2) * 2
        yoe_distribution[bucket] += 1

        exclude_reasons = []

        # ── Filter 1: country ─────────────────────────────────────────────────
        country_counter[country] += 1
        if country.lower() not in INDIA_VARIANTS:
            willing = c.get("redrob_signals", {}).get("willing_to_relocate", False)
            if not willing:
                non_india_ids.append((cid, country))
                exclude_reasons.append("non_india")

        # ── Filter 2: wrong title ─────────────────────────────────────────────
        title_counter[title] += 1
        if matches_any(title, WRONG_TITLE_KEYWORDS):
            wrong_title_ids.append((cid, title))
            exclude_reasons.append("wrong_title")

        # ── Filter 3: years of experience extremes ────────────────────────────
        if yoe < MIN_YOE:
            too_junior_ids.append((cid, yoe))
            exclude_reasons.append("too_junior")
        elif yoe > MAX_YOE:
            too_senior_ids.append((cid, yoe))
            exclude_reasons.append("too_senior")

        # ── Filter 4: null / empty skills ────────────────────────────────────
        if not skills:
            null_skills_ids.append(cid)
            exclude_reasons.append("null_skills")

        # ── Filter 5: honeypot skill signal ───────────────────────────────────
        if skills:
            expert_skills = [s for s in skills if s.get("proficiency") == "expert"]
            if expert_skills and all(
                (s.get("duration_months") or 0) == 0 for s in expert_skills
            ):
                honeypot_skill_ids.append(
                    (cid, [(s["name"], s.get("duration_months")) for s in expert_skills])
                )
                exclude_reasons.append("honeypot_skill")

        # ── Filter 6: pure consulting career ─────────────────────────────────
        # count firm appearances regardless (for the frequency report)
        for role in career:
            company = role.get("company", "")
            if is_consulting_firm(company):
                consulting_firm_counter[company] += 1

        if is_pure_consulting_career(career):
            companies = [r.get("company", "") for r in career]
            pure_consulting_ids.append((cid, companies))
            exclude_reasons.append("pure_consulting")

        if not exclude_reasons:
            passed.append(cid)
# inside your main with open(jsonl_path) loop, after career is loaded

# ── Filter 7: job hopper ──────────────────────────────────────────────────────
hopper, hopper_reason = is_job_hopper(career)
if hopper:
    job_hopper_ids.append((cid, hopper_reason))
    exclude_reasons.append("job_hopper")

# ── Filter 8: title chaser ────────────────────────────────────────────────────
chaser, chaser_reason = is_title_chaser(career)
if chaser:
    title_chaser_ids.append((cid, chaser_reason))
    exclude_reasons.append("title_chaser")
# ── Results ───────────────────────────────────────────────────────────────────

print("=" * 60)
print(f"TOTAL CANDIDATES: {total:,}")
print("=" * 60)

# YOE distribution
print("\n── YEARS OF EXPERIENCE DISTRIBUTION ──")
print(f"  {'YOE bucket':<15} {'count':>7}   bar")
for bucket in sorted(yoe_distribution):
    cnt = yoe_distribution[bucket]
    bar = "█" * (cnt // 200)
    label = f"{bucket}-{bucket+1}"
    marker = " ← MIN cutoff" if bucket == int(MIN_YOE // 2) * 2 else \
             " ← MAX cutoff" if bucket == int(MAX_YOE // 2) * 2 else ""
    print(f"  {label:<15} {cnt:>7,}   {bar}{marker}")
print(f"\n  Too junior (< {MIN_YOE} yoe):  {len(too_junior_ids):,}")
print(f"  Too senior (> {MAX_YOE} yoe): {len(too_senior_ids):,}")

# Null skills
print(f"\n── NULL / EMPTY SKILLS ──")
print(f"  Candidates with no skills listed: {len(null_skills_ids):,}")

# Honeypot skills
print(f"\n── HONEYPOT SKILL SIGNAL ──")
print(f"  Candidates with ALL expert skills at duration_months=0: {len(honeypot_skill_ids):,}")
print("  Sample (first 10):")
for cid, skills_info in honeypot_skill_ids[:10]:
    print(f"    {cid}  →  {skills_info}")

# Consulting firms
print(f"\n── CONSULTING FIRM FREQUENCY (all appearances in career history) ──")
for firm, cnt in consulting_firm_counter.most_common():
    print(f"  {firm:<40} {cnt:>6,}")
print(f"\n  Candidates with PURE consulting career: {len(pure_consulting_ids):,}")
print("  Sample (first 10):")
for cid, companies in pure_consulting_ids[:10]:
    print(f"    {cid}  →  {companies}")

# Combined
excluded_ids = (
    set(cid for cid, _ in non_india_ids)
    | set(cid for cid, _ in wrong_title_ids)
    | set(cid for cid, _ in too_junior_ids)
    | set(cid for cid, _ in too_senior_ids)
    | set(null_skills_ids)
    | set(cid for cid, _ in honeypot_skill_ids)
    | set(cid for cid, _ in pure_consulting_ids)
)

print(f"\n── COMBINED EXCLUSION SUMMARY ──")
print(f"  Non-India (no relocation):          {len(non_india_ids):>6,}")
print(f"  Wrong title:                        {len(wrong_title_ids):>6,}")
print(f"  Too junior (< {MIN_YOE} yoe):              {len(too_junior_ids):>6,}")
print(f"  Too senior (> {MAX_YOE} yoe):             {len(too_senior_ids):>6,}")
print(f"  Null/empty skills:                  {len(null_skills_ids):>6,}")
print(f"  Honeypot skill signal:              {len(honeypot_skill_ids):>6,}")
print(f"  Pure consulting career:             {len(pure_consulting_ids):>6,}")
print(f"  ─────────────────────────────────────────")
print(f"  Union (unique excluded):            {len(excluded_ids):>6,}  ({len(excluded_ids)/total*100:.1f}%)")
print(f"  Remaining after ALL filters:        {len(passed):>6,}")

print(f"  Job hopper:                         {len(job_hopper_ids):>6,}")
print(f"  Title chaser:                       {len(title_chaser_ids):>6,}")
```
